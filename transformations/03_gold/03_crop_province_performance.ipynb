# 03 - Crop Province Performance (Gold)

**Grain:** one row per province, crop, year and quarter. This is the most detailed table
for the dashboard and it keeps the `risk_label`.

```python
from pyspark.sql import functions as F

from common.delta_io import write_delta_table
from configs.etl_config import SILVER_CROP_WEATHER_TABLE, GOLD_CROP_PERFORMANCE_TABLE


enriched_df = spark.table(SILVER_CROP_WEATHER_TABLE)

performance_df = enriched_df.select(
    "province",
    "crop",
    "crop_group",
    "year",
    "quarter",
    "production",
    F.col("growth_rate_yoy").alias("production_yoy"),
    "production_lag_1",
    "production_lag_4",
    "climate_stress",
    "risk_label",
)

write_delta_table(performance_df, GOLD_CROP_PERFORMANCE_TABLE)
print(f"Rows: {spark.table(GOLD_CROP_PERFORMANCE_TABLE).count():,}")
```
